```python
import requests
import pandas as pd
```

## 出走表とタイムテーブル
朝9時までに用意想定.
 - 返却値
   - {"message": "OK", "data": {"timetable": json_data, "runtable": json_data}}

```python
url = "https://172.192.40.114/data"

headers = {
    # 認証キー:当日データAPIを使う時は共通
    "api-key": "AI_Keiba_2026",
    # "Racecards" or "Odds"
    "type_of_data": "Racecards",
    # Racecardsのときは日付, OddsのときはYYYYMMDDJJRRを入力:JJは場所コード, RRはレース番号
    # 場所コード："札幌", "函館", "福島", "新潟", "東京", "中山", "中京", "京都", "阪神", "小倉"は順番に01, 02, ..., 10
    "id": "20260712"
}
```

```python
# --- リクエスト送信 ---
# 形だけ証明書なのでverify=Falseにする: Warningは出るが無視でオッケー
response = requests.get(url, headers=headers, verify=False)

# --- 結果表示 ---
if response.status_code == 200:
    data = response.json()
    print("成功:", data["message"])
else:
    print(f"エラー: {response.status_code}")
    print(response.text)
```

```python
check = pd.DataFrame(data["data"]["timetable"])
check
```

```python
check = pd.DataFrame(data["data"]["runtable"])
check.head(11)
```
